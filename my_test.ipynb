```python
from core import *
import pandas as pd
import ast

def evaluate_micro(df, col):
    dix = {}
    true_pos = []
    false_pos = []
    false_neg = []  
    
    for i in range(len(df)):
        if len(df[col].iloc[i]) == 0:
            true_pos.append(0)
            false_pos.append(0)
            false_neg.append(len(df['curated'].iloc[i]))

        else:
            correct_i = 0 
            incorrect_i = 0 
            for x in df[col].iloc[i]:
                if x in df['curated'].iloc[i]:
                    correct_i = correct_i + 1
                else:
                    incorrect_i = incorrect_i + 1
                    
            
            true_pos.append(correct_i)
            false_pos.append(incorrect_i)
            false_neg.append( len(df['curated'].iloc[i]) - correct_i)


       
    df['TP'+"_"+col] = true_pos
    df['FP'+"_"+col] = false_pos
    df['FN'+"_"+col] = false_neg

    total_tp = df["TP"+"_"+col].sum()
    total_fp = df["FP"+"_"+col].sum()
    total_fn = df["FN"+"_"+col].sum()

    # Micro Precision, Recall, and F1-Score
    micro_precision = total_tp / (total_tp + total_fp)
    micro_recall = total_tp / (total_tp + total_fn)
    micro_f1_score = 2 * (micro_precision * micro_recall) / (micro_precision + micro_recall)

   

    dix['precision'] = micro_precision
    dix['recall'] = micro_recall
    dix['f1_score'] = micro_f1_score


    
    return dix
```
